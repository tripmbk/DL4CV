```python
import json
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import torch
import torch.nn as nn
import torchvision.transforms as T
from sklearn.metrics import confusion_matrix
from torch.optim import Adam
from torch.utils.data import DataLoader
from torchvision.datasets import MNIST
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)
```

```python
class RowTokenizer(nn.Module):

    def __init__(self, img_size=(32, 32), n_channels=1, d_model=64):
        super().__init__()
        H, W = img_size
        # Chiếu mỗi hàng (W * n_channels) sang d_model
        self.proj = nn.Linear(W * n_channels, d_model)

    def forward(self, x):
        # x: (B, C, H, W) -> (B, H, W * C)
        B, C, H, W = x.shape
        x = x.permute(0, 2, 3, 1).reshape(B, H, W * C)
        x = self.proj(x)  # (B, H, d_model)
        return x
```

```python
class MultiHeadAttention(nn.Module):

    def __init__(self, d_model, n_heads):
        super().__init__()
        self.n_heads = n_heads
        self.head_dim = d_model // n_heads
        self.qkv_proj = nn.Linear(d_model, 3 * d_model)
        self.out_proj = nn.Linear(d_model, d_model)

    def forward(self, x):
        B, N, C = x.shape
        qkv = self.qkv_proj(x).reshape(B, N, 3, self.n_heads, self.head_dim)
        qkv = qkv.permute(2, 0, 3, 1, 4)
        Q, K, V = qkv[0], qkv[1], qkv[2]

        attn = (Q @ K.transpose(-2, -1)) * (self.head_dim**-0.5)
        attn = torch.softmax(attn, dim=-1)
        out = (attn @ V).transpose(1, 2).reshape(B, N, C)
        return self.out_proj(out)
```

```python
class TransformerEncoder(nn.Module):

    def __init__(self, d_model, n_heads, r_mlp=4):
        super().__init__()
        self.ln1 = nn.LayerNorm(d_model)
        self.mha = MultiHeadAttention(d_model, n_heads)
        self.ln2 = nn.LayerNorm(d_model)
        self.mlp = nn.Sequential(
            nn.Linear(d_model, d_model * r_mlp),
            nn.GELU(),
            nn.Linear(d_model * r_mlp, d_model),
        )

    def forward(self, x):
        x = x + self.mha(self.ln1(x))
        x = x + self.mlp(self.ln2(x))
        return x
```

```python
class VisionTransformer(nn.Module):

    def __init__(
        self,
        d_model,
        n_classes,
        img_size,
        n_channels,
        n_heads,
        n_layers,
    ):
        super().__init__()
        H, W = img_size
        self.n_rows = H
        # Tổng số token = số hàng (H) + 1 token CLS
        self.max_seq_length = self.n_rows + 1

        self.row_embedding = RowTokenizer(img_size, n_channels, d_model)

        self.cls_token = nn.Parameter(torch.randn(1, 1, d_model))
        # Positional encoding cho toàn bộ chuỗi (CLS token + H hàng)
        self.pos_embedding = nn.Parameter(
            torch.randn(1, self.max_seq_length, d_model)
        )

        self.transformer_encoder = nn.Sequential(
            *[TransformerEncoder(d_model, n_heads) for _ in range(n_layers)]
        )
        self.classifier = nn.Linear(d_model, n_classes)

    def forward(self, images):
        B = images.shape[0]

        # 1. Biến đổi ảnh thành danh sách các row tokens: (B, H, d_model)
        x = self.row_embedding(images)

        # 2. Thêm CLS Token vào đầu chuỗi: (B, 1 + H, d_model)
        cls_tokens = self.cls_token.expand(B, -1, -1)
        x = torch.cat((cls_tokens, x), dim=1)

        # 3. Cộng Positional Embedding chuẩn cho toàn bộ chuỗi
        x = x + self.pos_embedding

        # 4. Đưa qua Transformer Encoder & Classifier
        x = self.transformer_encoder(x)
        return self.classifier(x[:, 0])  # Lấy đại diện từ CLS token
```

```python
d_model = 64
n_classes = 10
img_size = (32, 32)
n_channels = 1
n_heads = 8
n_layers = 6
alpha = 0.001
```

```python
transform = T.Compose([T.Resize(img_size), T.ToTensor()])
train_set = MNIST(
    root="./../datasets", train=True, download=True, transform=transform
)
test_set = MNIST(
    root="./../datasets", train=False, download=True, transform=transform
)

train_loader = DataLoader(train_set, shuffle=True, batch_size=batch_size)
test_loader = DataLoader(test_set, shuffle=False, batch_size=batch_size)
```

```python
transformer = VisionTransformer(
    d_model=d_model,
    n_classes=n_classes,
    img_size=img_size,
    n_channels=n_channels,
    n_heads=n_heads,
    n_layers=n_layers,
).to(device)

# Bộ tối ưu & Hàm mất mát
optimizer = Adam(transformer.parameters(), lr=alpha)
criterion = nn.CrossEntropyLoss()
```

```python
history = {
    "train_loss": [],
    "test_loss": [],
    "test_accuracy": [],
    "confusion_matrix": [],
}
epochs = 10
# VÒNG LẶP HUẤN LUYỆN & ĐÁNH GIÁ
for epoch in range(epochs):
    # Train Phase
    transformer.train()
    running_train_loss = 0.0
    for inputs, labels in train_loader:
        inputs, labels = inputs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = transformer(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_train_loss += loss.item()

    epoch_train_loss = running_train_loss / len(train_loader)
    history["train_loss"].append(epoch_train_loss)

    # Test Phase
    transformer.eval()
    running_test_loss = 0.0
    correct, total = 0, 0
    all_preds, all_labels = [], []

    with torch.no_grad():
        for inputs, labels in test_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = transformer(inputs)
            loss = criterion(outputs, labels)
            running_test_loss += loss.item()

            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

            # Thu thập nhãn cho epoch cuối để tính Confusion Matrix
            if epoch == epochs - 1:
                all_preds.extend(predicted.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())

    epoch_test_loss = running_test_loss / len(test_loader)
    epoch_accuracy = 100 * correct / total

    history["test_loss"].append(epoch_test_loss)
    history["test_accuracy"].append(epoch_accuracy)

    print(
        f"Epoch {epoch+1}/{epochs} - Train Loss: {epoch_train_loss:.4f} | Test Loss: {epoch_test_loss:.4f} | Accuracy: {epoch_accuracy:.2f}%"
    )

# Tính Confusion Matrix cho epoch cuối
cm = confusion_matrix(all_labels, all_preds)
history["confusion_matrix"] = cm.tolist()

# --- LƯU THÔNG TIN VÀO FILE JSON ---
json_path = "training_results_RowTokenizer.json"
with open(json_path, "w") as f:
    json.dump(history, f, indent=4)
print(f"\nĐã lưu kết quả huấn luyện vào file: {json_path}")

# --- VẼ ĐỒ THỊ BIỂU DIỄN ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Đồ thị 1: Training & Test Loss
axes[0].plot(
    range(1, epochs + 1), history["train_loss"], label="Train Loss", marker="o"
)
axes[0].plot(
    range(1, epochs + 1), history["test_loss"], label="Test Loss", marker="s"
)
axes[0].set_title("Loss over Epochs")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].legend()
axes[0].grid(True)

# Đồ thị 2: Test Accuracy
axes[1].plot(
    range(1, epochs + 1),
    history["test_accuracy"],
    color="green",
    marker="^",
    label="Test Accuracy",
)
axes[1].set_title("Accuracy over Epochs")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Accuracy (%)")
axes[1].legend()
axes[1].grid(True)

# Đồ thị 3: Confusion Matrix
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    ax=axes[2],
    xticklabels=range(10),
    yticklabels=range(10),
)
axes[2].set_title("Confusion Matrix (Final Epoch)")
axes[2].set_xlabel("Predicted Label")
axes[2].set_ylabel("True Label")

plt.tight_layout()
plt.savefig("training_metrics_RowTokenizer.png")
plt.show()
```
